```python
import numpy as np
import matplotlib.pyplot as plt

# Criar dados
x = np.linspace(0, 20, 100)
y = (x - 10)**2

# Encontrar o menor valor
indice_minimo = np.argmin(y)

print("Valor de x:", x[indice_minimo])
print("Valor mínimo de y:", y[indice_minimo])

# Criar gráfico
plt.plot(x, y, label="y = (x - 10)²")
plt.scatter(x[indice_minimo], y[indice_minimo], color="red")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Função quadrática")
plt.grid(True)
plt.legend()
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Criar os dados
dados = {
    "Horas_de_estudo": [1, 2, 3, 4, 5, 6, 7, 8],
    "Nota": [2, 3, 4, 5, 6, 7, 8, 9]
}

# Criar a tabela
df = pd.DataFrame(dados)

# Exibir a tabela
display(df)
```

```python

X = df[["Horas_de_estudo"]]
y = df["Nota"]

modelo = LinearRegression()
modelo.fit(X, y)

previsao = modelo.predict(
    pd.DataFrame({"Horas_de_estudo": [9]})
)

print("Nota prevista para 9 horas de estudo:", previsao[0])
```
